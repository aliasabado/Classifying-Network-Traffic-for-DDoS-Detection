```python
import numpy as np 
import pandas as pd 
from scipy.io import arff
```

```python
data = arff.loadarff('final-dataset.arff')
df = pd.DataFrame(data[0])
```

```python
columns = ['PKT_TYPE', 'FLAGS', 'NODE_NAME_FROM', 'NODE_NAME_TO', 'PKT_CLASS']
for c in columns:
    df[c] = df[c].str.decode('utf-8')
```

```python
labels = df['PKT_CLASS'].unique()
```

```python
# extremely impabalanced data
for i in labels:
    count = len(df[df['PKT_CLASS'] == i])
    print(f'{i}: {count}')
```

```python
df1 = df[df['PKT_CLASS'] != 'Normal']
df2 = df[df['PKT_CLASS'] == 'Normal'].sample(225000)
smaller_data = pd.concat([df1, df2], ignore_index=True)
output = smaller_data.sample(frac=1).reset_index(drop=True)
```

```python
output.to_csv('sample-data.csv')
```
